```python
import pandas as pd


df=pd.read_csv('tested.csv')
print(df.shape)
print(df.head())
```

```python
# data cleaning - use to simple
import pandas as pd
```

```python
# how to see missing values
print(df.isnull().sum())
```

```python
# some have a missing value age.i was watching average to fill
avg_avg=df['Age'].mean()
df['Age']=df['Age'].fillna(avg_avg)
```

```python
# in cabin had a numerious of missing values so i just removed
df=df.drop('Cabin',axis=1)
```

```python
# re-watching
print("after complete cleaning")
print(df.isnull().sum())
print(df.shape)
```

```python
df['Fare']=df['Fare'].fillna(df['Fare'].mean())
print(df.isnull().sum())
```

```python
# my first graph - sex count
import matplotlib.pyplot as plt

```

```python
# now watch male female count
count = df['Sex'].value_counts()
print(count)
```

```python
# now do a bar graph
count.plot(kind = 'bar')
plt.title("male vs female")
plt.show()
```
